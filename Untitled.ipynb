```python
import pandas as pd 
import numpy as np 
```

```python
df  = pd.read_csv("Countries.csv")

```

```python
# 1 which country has the highest population 
# we can use the sort menthond and find the first row using the head  using the ascending false hor the 

highest  = df.sort_values(
    by = "population",
    ascending = False 
)

highest["country"].head(1)

```

```python
# 2 what is the capital of the country with the hight populations
highest["capital_city"].head(1)
```

```python
# 3 which country have lowes populations using the min function 
lowest = df.sort_values(
    by = "population",
    ascending = True  
).head(1)
lowest["country"]
```

```python
# 4 whats the capital of the country with the least population
lowest["capital_city"]
```

```python
# 5 give me the top 5 highest county with the democratic score 
highest_democratic  = df.sort_values(
    by = "democracy_score",
    ascending = False 
).head(5)

highest_democratic["country"]
```
