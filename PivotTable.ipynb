```python
# What is a Pivot Table?
# A Pivot Table summarizes data by arranging values into rows and columns.

# For example, imagine your manager gives you sales data and asks:
# - What are total sales in each region?
# - How much does each category sell in each region?
# - Which region generates the highest revenue?
# - What is the average profit for each category?
# Instead of writing separate queries for every question, you can use a pivot table.

import pandas as pd

data = {
    "OrderID": [101, 102, 103, 104, 105, 106, 107, 108],
    "Region": [
        "North", "South", "North", "East",
        "South", "East", "North", "West"
    ],
    "Category": [
        "Electronics", "Furniture", "Furniture",
        "Electronics", "Electronics", "Furniture",
        "Electronics", "Furniture"
    ],
    "Sales": [5000, 3000, 7000, 8000, 4000, 6000, 9000, 2000],
    "Profit": [1000, 500, 1500, 2000, 800, 1200, 2500, 300]
}

df = pd.DataFrame(data)

print(df)
```

```python
pivot = pd.pivot_table (
    df,
    values = ["Profit","Sales"],
    index = ["Region","Category"],
    aggfunc = "sum",
    margins=True,
    fill_value=0,
    
).reset_index()

pivot


# - index="Region" creates row groups.
# - columns="Category" creates separate category columns.
# - values="Sales" selects the sales numbers.
# - aggfunc="sum" calculates the totals.
```

```python
# pd.crosstab() — Cross Tabulation
# pd.crosstab() is used to calculate the frequency or count of combinations between two or more categorical columns.

# It is especially useful for Data Analysts when analysing customer segments, sales categories, regions, departments and order statuses.
import pandas as pd

data = {
    "OrderID": [101, 102, 103, 104, 105, 106, 107, 108],
    "Region": [
        "North", "South", "North", "East",
        "South", "East", "North", "West"
    ],
    "Category": [
        "Electronics", "Furniture", "Furniture",
        "Electronics", "Electronics", "Furniture",
        "Electronics", "Furniture"
    ],
    "Sales": [5000, 3000, 7000, 8000, 4000, 6000, 9000, 2000],
    "Profit": [1000, 500, 1500, 2000, 800, 1200, 2500, 300]
}

bf = pd.DataFrame(data)
```

```python
bf
```

```python
result = pd.crosstab(
    bf["Region"],
    bf["Category"],
    margins = True,
)

result
```

```python
per = pd.crosstab (
    bf["Region"],
    bf["Category"],
    normalize = "columns"
)*100

per.round(1)

```

```python
r = pd.crosstab(
    bf["Region"],
    bf["Category"],
    values=bf["Sales"],
    aggfunc="sum"
).fillna(0)
```

```python
result = pd.crosstab(
    index=df["Region"],
    columns=df["Category"],
    values=df["Sales"],
    aggfunc="sum"
).fillna(0)

print(result)
```
