```python
import pandas as pd

customers = pd.DataFrame({
    "CustomerID": [1, 2, 3],
    "Name": ["Rahul", "Amit", "Priya"]
})

orders = pd.DataFrame({
    "OrderID": [101, 102, 103, 104],
    "CustomerID": [1, 2, 1, 4],
    "Amount": [500, 800, 300, 900]
})
```

```python
orders
```

```python
customers
```

```python
#merge() — Combine using a common column
# This is one of the most important methods for data analysis.
result = pd.merge(
    customers,orders,
    on = "CustomerID",
    how = "outer" )
```

```python
result
```

```python
# concat() — Stack DataFrames
# Use concat() when you want to combine datasets vertically or horizontally.
# Example A: Vertical concatenation (axis=0)

jan_sales = pd.DataFrame({
    "Product": ["Laptop", "Phone"],
    "Sales": [50000, 30000],
    "c":[1,2]
})

feb_sales = pd.DataFrame({
    "Product": ["Tablet", "Laptop"],
    "Sales": [20000, 60000],
    "b":[1,2]
})
```

```python
Total_sales = pd.concat(
    [jan_sales,feb_sales],
    axis = 0 ,
    ignore_index = "Ture"
 )
```

```python
Total_sales
```

```python
# join() — Combine using indexes
# join() is particularly useful when your DataFrames have meaningful indexes.


employees = pd.DataFrame({
    "Name": ["Rahul", "Amit", "Priya"]
}, index=[1, 2, 3])

salaries = pd.DataFrame({
    "Salary": [30000, 40000, 50000,55000]
}, index=[1, 2, 3,4])

```

```python
df = employees.join(salaries,how="outer")
```
