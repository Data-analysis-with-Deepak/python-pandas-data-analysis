```python
import pandas as pd 

data = {
    "Employe_id" : [11,22,33,44,55],
    "Employe_Name" : [ "Deepak","Babu", "Sham", "Ram", "Monu"],
    "Employe_Department" : ["IT","HR","IT","SALES","IT"],
    "Salary": [35000, 30000, 45000, 50000, 40000],
    "Age": [24, 26, 25, 27, 23]
}

df = pd.DataFrame(data)

# df

# df.shape with you can see thet how many row and how many cloumn we have in our dataset

# df.columns it show the columns name and the data type 

# df.dtypes it show the every columns name and data type what type of data is that cantain 

# df.values is show the value we have in the dataset in the form of array 

# df.size it tell the size of that dataset columns * rows 

    
```
