```python
#  now we  create  a function for the data exratoin from the particular column .
def extract_episode(title):
    data = title.split("(")[1].split(")")[0]
    return int(data.split()[0])  

    # here we are using the exratc mthond using the split method in pyhton
```

```python
 # 2 make a new column for the time stamp 
#  we have to exrtact the time stamp form the title from all each row 
# using this function we can find the date form the string why we use the find here it will find the and start str loop from there  and range will 
# extract the range between the str from that and +1 will use to


def time_stamp(t):
    s = t.find(")") +1
    result  = ""
    for i in range(s,s+19):
        result += t[i]
    return result.strip()
        
   
    
    
```
