```python
# Operations in Pandas

import pandas as pd
import numpy as np

data = {
    "Employee": ["Rahul", "Amit", "Priya", "Deepak", "Neha", "Karan"],
    "Department": ["IT", "HR", "IT", "Sales", "HR", "Sales"],
    "Salary": [50000, 35000, 60000, 45000, 40000, 55000],
    "Bonus": [5000, 3000, 7000, 4000, 3500, 6000],
    "Experience": [3, 1, 4, 2, 2, 5]
}

df = pd.DataFrame(data)

print(df)
```

```python
# A. Addition

# # Business question: Calculate total salary including bonus.

df["Total_Salary"] = df["Salary"] + df["Bonus"]
df
```

```python
# Subtraction

df["salary_dd"]=df["Salary"]-5000
df
```

```python
# Multiplication

# Calculate annual salary:


df["Anual"]=df["Salary"]*12
df
```

```python
# Comparison operations
# Comparison operations check whether a condition is true or false.

# Example: Find employees earning more than ₹45,000.

df["Salary"]>40000
df
```

```python
# NOT (~)

# Find employees who do not belong to IT.

nott = df[
    ~(df["Department"] == "IT")
    ]
print(nott)

# Important: Use parentheses around each condition when combining Pandas conditions.
```

```python
# Filtering with isin()
# # Suppose you want only IT and Sales employees.

fill = df[
    df["Department"].isin(["IT","HR"])
    ]
fill
```

```python
# Filtering with between()
# Find employees with salaries between ₹40,000 and ₹55,000.

between = df[
    df["Salary"].between(40000,50000)
    ]
between  #By default, both boundary values are included.
```

```python
# Sorting operations
# Sort employees by salary from highest to lowest.

sort = df.sort_values(
    by = "Salary",
    ascending = False
)
sort
```

```python
# Sort by multiple columns:

sort = df.sort_values(
    by = ["Experience","Salary"],
    ascending = False
)
sort
```

```python
# Adding, deleting and renaming columns

df["Tex"] = df["Salary"]*0.10
df
```

```python
# Delete a column:
df = df.drop(columns =["Total_Salary"])
df
```
