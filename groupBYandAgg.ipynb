```python
# 1. What is GroupBy?
# groupby() divides your data into groups based on one or more columns.

# For example, imagine you have this sales dataset:


import pandas as pd

data = {
    "Employee": ["Rahul", "Amit", "Priya", "Rahul", "Amit", "Priya"],
    "Department": ["IT", "HR", "IT", "IT", "HR", "Sales"],
    "Sales": [5000, 3000, 7000, 4000, 2000, 6000]
}

df = pd.DataFrame(data)

print(df)
```

```python
# Total sales by department
df.groupby("Department")["Sales"].sum()
```

```python
# Average sales by department
df.groupby("Department")["Sales"].mean()
```

```python
# Number of sales records by department
df.groupby("Department")["Sales"].count()
```

```python
# Highest sale by department
df.groupby("Department")["Sales"].max()
```

```python
# Lowest sale by department
df.groupby("Department")["Sales"].min()
```

```python
# GroupBy multiple columns
# This is particularly important in real-world analysis.
# Suppose your manager asks:
# "Calculate total sales for each department and employee."

df.groupby(["Department", "Employee"])["Sales"].sum()
```

```python
result = df.groupby(
    ["Department", "Employee"]
)["Sales"].sum().reset_index()

print(result)
```

```python

# Multiple aggregations using agg()
# Now let's learn something more powerful.
# Suppose your manager asks you to calculate:
# - Total sales
# - Average sales
# - Highest sale
# - Lowest sale
# You could write four separate groupby queries, but Pandas provides agg().

result = df.groupby("Department")["Sales"].agg(
    ["sum", "mean", "max", "min", "count"]
)

print(result)
```

```python
result = df.groupby("Department").agg(
    total_sales = ("Sales","sum")
)
print(result)
```

```python
import pandas as pd

data = {
    "OrderID": [101, 102, 103, 104, 105, 106, 107, 108],
    "Region": ["North", "South", "North", "East", "South", "East", "North", "West"],
    "Category": ["Electronics", "Furniture", "Furniture", "Electronics", "Electronics", "Furniture", "Electronics", "Furniture"],
    "Sales": [5000, 3000, 7000, 8000, 4000, 6000, 9000, 2000],
    "Quantity": [2, 1, 3, 4, 2, 2, 5, 1],
    "Profit": [1000, 500, 1500, 2000, 800, 1200, 2500, 300]
}

d = pd.DataFrame(data)
```

```python
d
```

```python
Total_sales = d.groupby("Region").agg(
    total_sales = ("Sales","sum")
    )
print(Total_sales)
```

```python
avg = d.groupby("Category").agg(
    avg_profit = ("Sales" ,"sum")
)
print(avg)
```

```python
Max_sales = d.groupby("Region").agg(
    total_sales = ("Sales","max")
    )
print(Max_sales)
```

```python
Total_orders = d.groupby("Region").agg(
    total_sales = ("OrderID","count")
    )
print(Total_orders)
```

```python
Sum_Avg = d.groupby("Region").agg(
    total_sales = ("Sales","sum"),
    total_avg = ("Sales","mean")
    
    )
print(Sum_Avg)
```

```python
total_sales_by_region_category = d.groupby(["Region","Category"])["Sales"].sum()
print(total_sales_by_region_category)
```
