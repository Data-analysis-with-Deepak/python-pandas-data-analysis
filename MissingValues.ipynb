```python
import pandas as pd
import numpy as np

data = {
    "Employee": ["Rahul", "Amit", "Priya", "Deepak", "Neha"],
    "Age": [24, np.nan, 25, 27, np.nan],
    "Salary": [35000, 30000, np.nan, 50000, 40000],
    "Department": ["IT", "HR", "Sales", np.nan, "Finance"]
}

df = pd.DataFrame(data)

df
```

```python
# using this we can find the null value in every column and there count using the sum() we can find the count and isna() tell us 
# the there is null value have or not using true or flase
df.isna() # this is return the null value using true or false 
df.isna().sum() # this will return the null count in each column with there name 
df.isna().any() # this will return the is any column has the null value or not using ture or false 
df.isnull().sum().sum() # it will show the dataset have total null value
(df.isnull().mean()*100).round(2) #Find the percentage of missing values in each column
```

```python
# how to remove the null value in the dataset

df.dropna() # this will the remove the null value from the table or dataset and it will show the only the row which no have null value in the dataset
```

```python
df.dropna(thresh = 2)  # what it will do it remove the row which have the more then 2 not null value 
```

```python

#now we have to fill the null value in the data set
#this will replace all the null value with the 0 

# df.fillna(0)

# but we can work like that we have to fill the diffrent column with diffrent value so we create a demion and fill the value
# for the column nane and update

value = {
    'Employee':"Unknown",
    "Age" :"NAN",
    "Salary": "NAN",
    "Department":"Unknown"
}
df.fillna(value=value,inplace=True)
    
    
    
    
```
