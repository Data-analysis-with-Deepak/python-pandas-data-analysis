```python
# duplicated() — identify duplicate records.
# replace() — replace incorrect or unwanted values.
# apply() — apply custom functions to rows or columns.

import pandas as pd
import numpy as np

data = {
    "EmployeeID": [101, 102, 103, 104, 102, 106, 107],
    "Employee": ["Rahul", "Amit", "Priya", "Deepak", "Amit", "Neha", "Karan"],
    "Department": ["IT", "HR", "IT", "Sales", "HR", "hr", "Sales"],
    "Salary": [50000, 35000, 60000, 45000, 35000, 40000, 55000],
    "Status": ["Active", "active", "Inactive", "Active", "active", "Unknown", "Inactive"]
}

df = pd.DataFrame(data)

print(df)
```

```python
# duplicated() — Find duplicate records
# duplicated() identifies whether a row or value has appeared before.


df.duplicated()
```

```python
# df = df.drop(df.duplicated())
```

```python
# Find duplicates using a particular column
df["EmployeeID"].duplicated()
```

```python
dup = df[df["EmployeeID"].duplicated()]
dup
```

```python
# Check duplicates using multiple columns

df.duplicated(
    subset=["Employee","Department"],
    keep = False
)
```

```python
# Remove duplicates using drop_duplicates()
clean = df.drop_duplicates(
    subset=["EmployeeID"],
    keep = "first"
)
clean


# Important parameters:
# Parameter	Meaning
# subset	Columns to check
# keep="first"	Keep first occurrence
# keep="last"	Keep last occurrence
# keep=False	Remove every occurrence in a duplicate group

# Real-world example: If an order dataset contains duplicate OrderIDs, you may need to identify and 
# remove duplicate transactions before calculating revenue. But always investigate which record is
# valid before deleting anything.
    
```
